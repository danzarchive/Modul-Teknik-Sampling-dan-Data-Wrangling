```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from datetime import datetime
import time

def scrape_detik_news():
    base_url = "https://www.detik.com/search/searchall"  # Gunakan search page untuk struktur stabil
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
    }
    params = {
        'query': '',  # Kosong untuk latest news umum
        'sortby': 'time'
    }
    
    try:
        response = requests.get(base_url, headers=headers, params=params)
        soup = BeautifulSoup(response.text, 'html.parser')
        
        # Cari articles lebih luas (semua <article>, seperti di tutorial 2025)
        articles = soup.find_all('article')[:30]  # Ambil 30 untuk kompensasi
        
        news_data = []
        valid_count = 0
        for article in articles:
            # Cari title_elem: Prioritas h3.media__title, fallback h2
            title_elem = article.find('h3', class_='media__title') or article.find('h2', class_='media__title')
            if not title_elem:
                # Fallback lebih luas jika masih gak ada
                title_elem = article.find(['h2', 'h3'])
                if not title_elem:
                    continue
            
            # Ambil a_tag dari title_elem (spesifik untuk title link)
            a_tag = title_elem.find('a')
            if not a_tag:
                continue
            
            # Title dari a_tag text
            title = a_tag.text.strip()
            if not title:  # Skip jika title kosong
                continue
            
            # Link: Full URL
            link = a_tag['href']
            if link.startswith('/'):
                link = 'https://www.detik.com' + link
            
            # Publish time: Cari div.media__date, ambil text (include span.time__text jika ada)
            time_elem = article.find('div', class_='media__date')
            if time_elem:
                # Jika ada span di dalam, ambil dari situ; else text langsung
                time_span = time_elem.find('span', class_='time__text')
                publish_time = time_span.text.strip() if time_span else time_elem.text.strip()
            else:
                publish_time = 'N/A'
            
            # Summary
            summary_elem = article.find('div', class_='media__desc')
            summary = summary_elem.text.strip() if summary_elem else 'N/A'
            
            news_data.append({
                'title': title,
                'link': link,
                'publish_time': publish_time,
                'summary': summary,
                'scraped_at': datetime.now().isoformat()
            })
            
            valid_count += 1
            if valid_count >= 10:  # Stop setelah 10 valid
                break
        
        # Convert to DataFrame
        df = pd.DataFrame(news_data)
        return df
        
    except Exception as e:
        print(f"Error scraping Detik: {e}")
        return None

# Jalankan scraping
news_df = scrape_detik_news()
if news_df is not None:
    print(f"Berhasil scraping {len(news_df)} berita")
    print(news_df.head(10))  # Print 10 untuk lihat semua
    
    # Simpan ke CSV
    # news_df.to_csv('detik_news.csv', index=False, encoding='utf-8')
    news_df
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from datetime import datetime

def scrape_detik_news(query='berita', num_articles=10):
    base_url = "https://www.detik.com/search/searchall"
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
    }
    params = {
        'query': query,
        'sortby': 'time',
        'page': 1
    }
    
    try:
        response = requests.get(base_url, headers=headers, params=params)
        if response.status_code != 200:
            print(f"Error: Status code {response.status_code}")
            return None
        
        soup = BeautifulSoup(response.text, 'html.parser')
        articles = soup.find_all('article')
        
        news_data = []
        count = 0
        for article in articles:
            if count >= num_articles:
                break
            
            # Title: Cari h2
            title_elem = article.find('h2')
            title = title_elem.get_text(strip=True) if title_elem else 'N/A'
            if title == 'N/A':
                continue
            
            # Link: Cari a href=True pertama (dari tutorial)
            link_elem = article.find('a', href=True)
            link = link_elem['href'] if link_elem else 'N/A'
            if link.startswith('/'):
                link = 'https://www.detik.com' + link
            
            # Publish time
            time_elem = article.find('div', class_='media__date')
            publish_time = time_elem.get_text(strip=True) if time_elem else 'N/A'
            
            # Summary
            summary_elem = article.find('div', class_='media__desc')
            summary = summary_elem.get_text(strip=True) if summary_elem else 'N/A'
            
            news_data.append({
                'title': title,
                'link': link,
                'publish_time': publish_time,
                'summary': summary,
                'scraped_at': datetime.now().isoformat()
            })
            
            count += 1
        
        df = pd.DataFrame(news_data)
        return df
        
    except Exception as e:
        print(f"Error scraping Detik: {e}")
        return None

# Jalankan scraping
news_df = scrape_detik_news(query='berita', num_articles=10)
if news_df is not None:
    print(f"Berhasil scraping {len(news_df)} berita")
    print(news_df.head(10))
    
    # Simpan ke CSV
    # news_df.to_csv('detik_news.csv', index=False, encoding='utf-8')
    news_df
else:
    print("Gagal scraping, coba ganti query (misal 'politik' atau 'olahraga')")
```

```python
import requests
import pandas as pd
from datetime import datetime
import json

def fetch_bmkg_weather_api(adm4_code):
    url = f"https://api.bmkg.go.id/publik/prakiraan-cuaca?adm4={adm4_code}"
    headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36'}
    
    try:
        response = requests.get(url, headers=headers)
        response.raise_for_status()
        data = response.json()
        
        # Asumsi data ada di 'data' key, ambil forecast pertama (sesuaikan jika perlu)
        if 'data' in data and data['data']:
            forecast = data['data'][0]  # Ambil hari pertama
            location_info = data.get('lokasi', {})  # Info lokasi
            
            return {
                'city': f"{location_info.get('desa', 'Unknown')}, {location_info.get('kecamatan', '')}",
                'weather_condition': forecast.get('weather_desc', 'N/A'),
                'temperature': f"{forecast.get('t', 'N/A')}°C",
                'humidity': f"{forecast.get('hu', 'N/A')}%",
                'scraped_at': datetime.now().isoformat()
            }
        else:
            print("Data tidak ditemukan")
            return None
    except Exception as e:
        print(f"Error API BMKG: {e}")
        return None

# Contoh penggunaan: Kode wilayah untuk Jakarta Pusat (ganti dengan kode lain)
adm4_code = "31.71.03.1001"  # Kelurahan Kemayoran, Jakarta Pusat
weather_data = fetch_bmkg_weather_api(adm4_code)

if weather_data:
    df = pd.DataFrame([weather_data])
    print(f"Data cuaca untuk {weather_data['city']}")
    print(df)
else:
    print("Gagal mengambil data")
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup
import pandas as pd
from datetime import datetime

options = Options()
options.headless = True
driver = webdriver.Chrome(options=options)
driver.get("https://www.bmkg.go.id/cuaca/prakiraan-cuaca-indonesia.bmkg")
soup = BeautifulSoup(driver.page_source, 'html.parser')

# Cari elemen cuaca (sesuaikan selector berdasarkan inspect)
weather_cards = soup.find_all('div', class_='weather-item')  # Contoh; ganti dengan class sebenarnya
weather_data = []
for card in weather_cards:
    city = card.find('h3', class_='city-name').text.strip() if card.find('h3', class_='city-name') else 'N/A'
    condition = card.find('.condition').text.strip() if card.find('.condition') else 'N/A'
    temp = card.find('.temp').text.strip() if card.find('.temp') else 'N/A'
    hum = card.find('.humidity').text.strip() if card.find('.humidity') else 'N/A'
    weather_data.append({'city': city, 'weather_condition': condition, 'temperature': temp, 'humidity': hum})

df = pd.DataFrame(weather_data)
print(df.head())
driver.quit()
```

```python
import requests
import pandas as pd
from datetime import datetime
import json

def fetch_bmkg_weather_api(adm4_code, location_name):
    url = f"https://api.bmkg.go.id/publik/prakiraan-cuaca?adm4={adm4_code}"
    headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36'}
    
    try:
        response = requests.get(url, headers=headers)
        response.raise_for_status()
        data = response.json()
        
        if 'data' in data and data['data']:
            forecast = data['data'][0]  # Ambil forecast pertama (terdekat)
            return {
                'city': location_name,
                'weather_condition': forecast.get('weather_desc', 'N/A'),
                'temperature': f"{forecast.get('t', 'N/A')}°C",
                'humidity': f"{forecast.get('hu', 'N/A')}%",
                'wind_speed': f"{forecast.get('ws', 'N/A')} km/jam",
                'datetime': forecast.get('local_datetime', 'N/A'),
                'scraped_at': datetime.now().isoformat()
            }
        else:
            print(f"Tidak ada data untuk {location_name}")
            return None
    except Exception as e:
        print(f"Error untuk {location_name}: {e}")
        return None

# Daftar kota besar dengan kode adm4 (contoh; tambah dari CSV jika perlu)
cities = {
    "31.71.03.1001": "Kemayoran, Jakarta Pusat",  # Jakarta
    # "32.21.01.1001": "Cicendo, Bandung",           # Bandung
    # "35.54.07.1001": "Bulak, Surabaya",            # Surabaya
    "12.71.02.1001": "Medan Maimun, Medan",        # Medan
    "73.11.01.1001": "Maros, Makassar"             # Makassar
}

weather_data = []
for code, name in cities.items():
    result = fetch_bmkg_weather_api(code, name)
    if result:
        weather_data.append(result)

df = pd.DataFrame(weather_data)
if not df.empty:
    print(f"Data cuaca untuk {len(df)} kota (diperbarui {datetime.now().strftime('%Y-%m-%d %H:%M')})")
    print(df[['city', 'weather_condition', 'temperature', 'humidity']])
else:
    print("Gagal mengambil data untuk semua kota")
```
